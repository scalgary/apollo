```python
# Test 2 : Créer un user avec un membership

# Étape 1 : Créer un event_type (nécessaire pour le membership)
event_type = create_event_type(name="open_play", display_name="Intérieur")
print(f"✅ EventType créé : {event_type.name}, ID: {event_type.id}")

# Étape 2 : Créer un user
user = create_user(email="alice@test.com", display_name="Alice")
print(f"✅ User créé : {user.email}, ID: {user.id}")

# Étape 3 : Créer un membership pour ce user
membership = create_membership(  # ← PAS _helper
    user_id=user.id,
    event_type_id=event_type.id,
    membership_type="full_member"
)
print(f"✅ Membership créé : ID {membership.id}")

# Étape 4 : Vérifier dans la DB
from db_models import UserEventTypeMembership
db_membership = db.query(UserEventTypeMembership).filter_by(
    user_id=user.id,
    event_type_id=event_type.id
).first()

# Assertions
assert db_membership is not None, "Le membership devrait exister"
assert db_membership.membership_type == "full_member"
assert db_membership.user_id == user.id
assert db_membership.event_type_id == event_type.id

print("✅ Test réussi : Alice a un membership full_member pour open_play")
```
