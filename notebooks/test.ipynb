```python
# Cellule 1 : Test de connexion basique
from sqlalchemy import create_engine, text

# URL de connexion (même que dans docker-compose)
DATABASE_URL = "postgresql://apollo:apollo123@db:5432/apollo"

# Créer le moteur
engine = create_engine(DATABASE_URL)

# Tester la connexion
with engine.connect() as conn:
    result = conn.execute(text("SELECT version()"))
    postgres_version = result.fetchone()[0]
    print("✓ Connexion PostgreSQL réussie !")
    print(f"Version PostgreSQL : {postgres_version}")
```
